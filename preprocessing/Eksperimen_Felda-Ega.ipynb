# Eksperimen Data: Breast Cancer Wisconsin Diagnostic
Deskripsi dataset: Klasifikasi biner untuk mendeteksi kanker payudara.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
```

```python
# Load dataset
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
```

## Exploratory Data Analysis (EDA)

```python
sns.countplot(x='target', data=df)
plt.title('Distribusi Target')
plt.show()
```

```python
# Heatmap korelasi
plt.figure(figsize=(10,8))
sns.heatmap(df.iloc[:, :10].corr(), annot=True, cmap='coolwarm')
plt.title('Korelasi 10 Fitur Pertama')
plt.show()
```

## Preprocessing

```python
# Hapus duplikasi jika ada
df = df.drop_duplicates()
```

```python
# Pisahkan fitur dan target
X = df.drop('target', axis=1)
y = df['target']
```

```python
# Scaling numerik
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
```

```python
# Gabungkan kembali
df_processed = pd.concat([X_scaled_df, y.reset_index(drop=True)], axis=1)
```

```python
# Simpan data preprocessing
df_processed.to_csv('breast_cancer_preprocessing.csv', index=False)
print('Data berhasil disimpan ke breast_cancer_preprocessing.csv')
```
